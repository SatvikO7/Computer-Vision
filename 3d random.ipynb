```python
import numpy as np
import matplotlib.pyplot as plt

N = 200
X = np.random.rand(N) * 10
Y = np.random.rand(N) * 10
Z = np.random.rand(N) * 10

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter(X, Y, Z)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
plt.title('Random 3D Points')

plt.show()
```
