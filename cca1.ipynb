```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

IMG_PATH = r"test2.jpg"
img = cv2.imread(IMG_PATH)                              # BGR
assert img is not None, "Image not found. Update IMG_PATH."

# Convert to RGB for display
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Applying linear filters
# higher the k value means more smoothing)
avg3 = cv2.blur(img_rgb, (3, 3))
avg7 = cv2.blur(img_rgb, (7, 7))

# Gaussian filter
gauss3 = cv2.GaussianBlur(img_rgb, (3, 3), 0)
gauss7 = cv2.GaussianBlur(img_rgb, (7, 7), 0)

#Compare 
plt.figure(figsize=(12, 8))

plt.subplot(2,3,1); plt.imshow(img_rgb); plt.title("Original (Noisy)"); plt.axis('off')
plt.subplot(2,3,2); plt.imshow(avg3);    plt.title("Average 3×3");       plt.axis('off')
plt.subplot(2,3,3); plt.imshow(avg7);    plt.title("Average 7×7");       plt.axis('off')
plt.subplot(2,3,5); plt.imshow(gauss3);  plt.title("Gaussian 3×3");      plt.axis('off')
plt.subplot(2,3,6); plt.imshow(gauss7);  plt.title("Gaussian 7×7");      plt.axis('off')

plt.tight_layout()
plt.show()

```
