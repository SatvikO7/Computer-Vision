```python
import numpy as np
import matplotlib.pyplot as plt
from imageio import imread

gray_img = imread('peppers.jpeg')

F = np.fft.fft2(gray_img)
F_shifted = np.fft.fftshift(F)

magnitude_gray = np.log(1 + np.abs(F_shifted))
phase_gray = np.angle(F_shifted)

plt.figure(figsize=(12,4))
plt.subplot(1,3,1); plt.imshow(gray_img, cmap='gray'); plt.title('Original Grayscale Image')
plt.subplot(1,3,2); plt.imshow(magnitude_gray, cmap='gray'); plt.title('Magnitude Spectrum')
plt.subplot(1,3,3); plt.imshow(phase_gray, cmap='gray'); plt.title('Phase Spectrum')
plt.show()


color_img = imread('peppers.jpeg') / 255.0

R = color_img[:,:,0]
G = color_img[:,:,1]
B = color_img[:,:,2]

FR = np.fft.fftshift(np.fft.fft2(R))
FG = np.fft.fftshift(np.fft.fft2(G))
FB = np.fft.fftshift(np.fft.fft2(B))

magR = np.log(1 + np.abs(FR))
magG = np.log(1 + np.abs(FG))
magB = np.log(1 + np.abs(FB))

plt.figure(figsize=(10,8))
plt.subplot(2,2,1); plt.imshow(color_img); plt.title('Original Color Image')
plt.subplot(2,2,2); plt.imshow(magR, cmap='gray'); plt.title('Red Channel Magnitude Spectrum')
plt.subplot(2,2,3); plt.imshow(magG, cmap='gray'); plt.title('Green Channel Magnitude Spectrum')
plt.subplot(2,2,4); plt.imshow(magB, cmap='gray'); plt.title('Blue Channel Magnitude Spectrum')
plt.show()
```
