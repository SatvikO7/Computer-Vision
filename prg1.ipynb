```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

# 1. Read an RGB image
img = cv2.imread(cv2.samples.findFile('peppers.jpeg'))  # built-in OpenCV test image

img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# 2. Convert to numeric matrix (like MATLAB's double)
img_matrix = img.astype(float)

# 3. Split into R, G, B channels
R = img[:, :, 0]
G = img[:, :, 1]
B = img[:, :, 2]

# 4. Create RGB images for each channel
zeros = np.zeros_like(R)
red_img   = np.stack((R, zeros, zeros), axis=2)
green_img = np.stack((zeros, G, zeros), axis=2)
blue_img  = np.stack((zeros, zeros, B), axis=2)

# 5. Display results using Matplotlib
plt.figure(figsize=(10, 10))

# Original Image
plt.subplot(3, 3, 2)
plt.imshow(img)
plt.title('Original RGB Image')
plt.axis('off')

# RGB representations
plt.subplot(3, 3, 4)
plt.imshow(red_img)
plt.title('Red Channel (RGB)')
plt.axis('off')

plt.subplot(3, 3, 5)
plt.imshow(green_img)
plt.title('Green Channel (RGB)')
plt.axis('off')

plt.subplot(3, 3, 6)
plt.imshow(blue_img)
plt.title('Blue Channel (RGB)')
plt.axis('off')

# Grayscale representations
plt.subplot(3, 3, 7)
plt.imshow(R, cmap='gray')
plt.title('Red Channel (Gray)')
plt.axis('off')

plt.subplot(3, 3, 8)
plt.imshow(G, cmap='gray')
plt.title('Green Channel (Gray)')
plt.axis('off')

plt.subplot(3, 3, 9)
plt.imshow(B, cmap='gray')
plt.title('Blue Channel (Gray)')
plt.axis('off')

plt.tight_layout()
plt.show()

# 6. Optionally save output images
os.makedirs("output_channels", exist_ok=True)

cv2.imwrite('output_channels/Red_Channel_RGB.png', cv2.cvtColor(red_img, cv2.COLOR_RGB2BGR))
cv2.imwrite('output_channels/Green_Channel_RGB.png', cv2.cvtColor(green_img, cv2.COLOR_RGB2BGR))
cv2.imwrite('output_channels/Blue_Channel_RGB.png', cv2.cvtColor(blue_img, cv2.COLOR_RGB2BGR))
cv2.imwrite('output_channels/Red_Channel_Gray.png', R)
cv2.imwrite('output_channels/Green_Channel_Gray.png', G)
cv2.imwrite('output_channels/Blue_Channel_Gray.png', B)

print("✅ All channel images saved inside 'output_channels' folder.")
```
