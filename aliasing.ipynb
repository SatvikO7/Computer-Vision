```python
import numpy as np
import matplotlib.pyplot as plt

cb = np.kron([[1, 0] * 4 + [1], [0, 1] * 4 + [0]] * 5, np.ones((10,10)))

cb_2 = cb[::2, ::2]
cb_4 = cb[::4, ::4]
cb_8 = cb[::8, ::8]

plt.figure(figsize=(10,10))
plt.subplot(2,2,1); plt.imshow(cb, cmap='gray'); plt.title('Original')
plt.subplot(2,2,2); plt.imshow(cb_2, cmap='gray'); plt.title('Sampled x2')
plt.subplot(2,2,3); plt.imshow(cb_4, cmap='gray'); plt.title('Sampled x4')
plt.subplot(2,2,4); plt.imshow(cb_8, cmap='gray'); plt.title('Sampled x8')
plt.show()
```
